```python

def save_to_csv(dic_out, request, file_name):
    date = list(dic_out[request['variable'][0]].keys())[0]
    df = pd.DataFrame(dic_out[request['variable'][0]][date])
    df.to_csv(file_name, index=False)
    print(f"Archivo '{file_name}' guardado exitosamente.")

# Listado de requests
requests = [
    request4,
    request5,
    request6,
    request7,
    request8,
    request9,
    request10,
    request11,
    request12,
    request13
]

# Nombres de archivo correspondientes
file_names = [
    'temperatura_del_aire.csv',
    'presion_atmosferica.csv',
    'precipitacion_acumulada.csv',
    'presion_barometrica.csv',
    'humedad_relativa.csv',
    'salinidad.csv',
    'nivel_del_mar.csv',
    'temperatura_del_mar.csv',
    'direccion_del_viento.csv',
    'velocidad_del_viento.csv'
]

# Guardar cada request en un archivo CSV correspondiente
for req, file_name in zip(requests, file_names):
    save_to_csv(dic_out, req, file_name)
```

**NOTAS:**

> Luego de consumir los datos, las fechas deben convertirse a formatos de fechas como tal, ya que, por estandar mediante HTTP, se hace uso del ISO 8601 (YYYY-MM-DDTHH:MM:SSZ), tanto en la petición como en la respuesta.

> Es necesario definir y ejecutar la API para consumir los datos de la bodega desde el contenedor de backend, dado que la bodega está implementada en PostgreSQL y no en una tecnología de scripting que permita su desarrollo dentro del propio contenedor (`data_wharehouse`). A fines prácticos, con la función `extract_data_from_data_warehouse` se puede hacer dicho consumo.
